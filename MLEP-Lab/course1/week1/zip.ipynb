```python
from pathlib import Path

folder = Path("images")

print("Files:")
for p in folder.rglob("*"):
    if p.is_file():
        print(p, p.stat().st_size)

print("\nTotal bytes:")
print(sum(p.stat().st_size for p in folder.rglob("*") if p.is_file()))
```

```python
from pathlib import Path
import zipfile

source = Path("images")
output = Path("test_images.zip")

with zipfile.ZipFile(output, "w", compression=zipfile.ZIP_STORED) as z:
    for p in source.iterdir():
        if p.is_file():
            z.write(p, p.name)

print("ZIP size:", output.stat().st_size, "bytes")
```

```python
from pathlib import Path
import zipfile

zip_path = Path("test_images.zip")

print("ZIP path:", zip_path.resolve())
print("ZIP size:", zip_path.stat().st_size / 1024**3, "GB")

with zipfile.ZipFile(zip_path, "r") as z:
    print("Number of entries:", len(z.infolist()))

    for info in z.infolist()[:20]:
        print(
            info.filename,
            "compressed:", info.compress_size,
            "uncompressed:", info.file_size
        )
```
